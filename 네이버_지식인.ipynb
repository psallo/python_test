```python
def naver_kin_one(page):
  from bs4 import BeautifulSoup
  from urllib import request
  import pandas as pd
  import datetime

  from selenium import webdriver
  import time

  con2 = request.urlopen('https://kin.naver.com/search/list.naver?query=%EC%B7%A8%EC%97%85&page=' + str(page))
  doc2 = BeautifulSoup(con2, 'html.parser')

  rows = []

  for i in range(1, 11):
    # 한페이지당 10개의 제목
    title_list = doc2.select('#s_content > div.section > ul > li:nth-child(' + str(i) + ') > dl > dt > a')
    print('제목:', title_list[0].text.strip())
    title = title_list[0].text.strip()

    # 날짜 추출
    date_list = doc2.select('#s_content > div.section > ul > li:nth-child(' + str(i) + ') > dl > dd.txt_inline')
    print('날짜:', date_list[0].text.strip())
    date = date_list[0].text.strip()

    # 내용 추출
    content_list = doc2.select('#s_content > div.section > ul > li:nth-child(' + str(i) + ') > dl > dd:nth-child(3)')
    print('내용:', content_list[0].text.strip())
    content = content_list[0].text.strip()

    row = [title, date, content]
    rows.append(row)
  print('-----------------------------------------------------------------------------------------------------------------------------------------page no.', page)

  return rows
```

```python
result_1 = naver_kin_one(200)
```
