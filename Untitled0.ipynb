람다 함수

```python
def call1(a):
  return a**2
```

```python
call1(10)
```

```python
print((lambda a : a**2)(10))
```

```python
a = [ 1, 2, 3]
result = map(call1, a)
print(list(result))
```

```python
a2 = [10, 20, 30]
result2 = map(lambda a: a**3, a2)
print(list(result2))
```

```python
a_fun = lambda a : a - 1
a_fun(10)
```
