```python
def get_finance_data(code):
    from bs4 import BeautifulSoup
    from urllib import request
    import pandas as pd
    import datetime
    con3 = request.urlopen('https://finance.naver.com/item/main.naver?code=' + code)
    doc3 = BeautifulSoup(con3, 'html.parser')

    # 회사 이름 추출
    title_list = doc3.select(".wrap_company>h2>a")
    title =title_list[0].text.strip()
    print(title)


    # 코드 추출
    code_list = doc3.select("span.code")
    code = code_list[0].text.strip()
    print(code)


    # 오늘가 추출
    price_list = doc3.select("div.today span.blind")
    price_1 = price_list[0].text.strip()
    price_today = int(price_1.replace(",",""))
    print(price_today)


    # 시작가 추출
    price_list2 = doc3.select("td.first span.blind")
    price_2 = price_list2[0].text.strip()
    price_yes = int(price_2.replace(",",""))
    print(price_yes)


    price_3 = price_list2[1].text.strip()
    price_high = int(price_3.replace(",",""))
    print(price_high)

    com = [code,title, price_today, price_yes, price_high]
    return com
```

```python
get_finance_data('005930')
```

```python
get_finance_data('035720')
```

```python
rows = []
```

```python
code_list = ['005930', '035720', '035420', '042700']
```

```python
for code in code_list:
    com = get_finance_data(code)
    rows.append(com)
    print("---------------------------------------------")
```

```python
rows
```

```python
df = pd.DataFrame(rows, columns=['code', 'title', 'price_today', 'price_yes', 'price_high'])
```
